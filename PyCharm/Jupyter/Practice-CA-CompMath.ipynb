```python
from sympy import Symbol
john=Symbol('john')
sen=Symbol('sen')
print(Matrix([[john,2,3], [1,sen,5]]))
```

```python
A=Matrix([[1,-3,2],[-3,3,-1], [2,-1,0]])
c=A.row_insert(1, Matrix([[0,0,1]]))
print(c)
```

```python
B=Matrix([[1,2,3],[2,3,4], [1,5,7]])
d=B.col_insert(1, Matrix([7,7,3]))
print(d)
```
