```python
artifacts = [12, 7, 99, 23, 7, 45, 99, 8, 4, 0]
```

```python
# Q1. A new artifact with value 88 was found — add it to the end.

artifacts.append(88)
print(artifacts)
```

```python
# Q2. A secret item of value 33 must be placed at index 2.

artifacts.insert(2, 33)
print(artifacts)
```

```python
# Q3. There's a rumor that the value 7 is cursed — remove the first occurrence.

artifacts.remove(7)
print(artifacts)
```

```python
# Q4. A traitor stole the last artifact — remove it without using the index.

artifacts.pop()  # no index given -> removes the last item
print(artifacts)
```

```python
# Q5. Another traitor took the first one — remove that one using its index.

artifacts.pop(0)
print(artifacts)
```

```python
# Q6. A new legend was found — what's the index of artifact 45?

print(artifacts.index(45))
```

```python
# Q7. The value 99 appears suspiciously often — how many times?

print(artifacts.count(99))
```

```python
# Q8. Sort the artifacts from smallest to largest.

artifacts.sort()
print(artifacts)
```

```python
# Q9. Reverse the list — the elder wants it in backward order.

artifacts.reverse()
print(artifacts)
```

```python
# Q10. Extract just the 3rd and 4th artifacts using slicing.

print(artifacts[2:4])
```

```python
# Q11. Extract every 2nd artifact from the list.

print(artifacts[::2])
```

```python
# Q12. Make a copy of the current list into a new variable.

artifacts_copy = artifacts.copy()
print(artifacts_copy)
```

```python
# Q13. Merge this new list with another batch: [101, 202, 303].

artifacts_copy.extend([101, 202, 303])
print(artifacts_copy)
```

```python
# Q14. Add the sum of all current artifact values to the list.

artifacts.append(sum(artifacts))
print(artifacts)
```

```python
# Q15. Add another entry: the maximum value in the list.

artifacts.append(max(artifacts))
print(artifacts)
```

```python
# Q16. What is the final length of the list?

print(len(artifacts))
```

```python
# Q17. What is the index of 0?

print(artifacts.index(0))
```

```python
# Q18. Find the maximum and minimum artifact values.

print(max(artifacts), min(artifacts))
```

```python
# Q19. Clear the copy list — the new batch was a decoy!

artifacts_copy.clear()
print(artifacts_copy)
```

```python
# Q20. Check if the number 23 is still in the original list.

print(23 in artifacts)
```
