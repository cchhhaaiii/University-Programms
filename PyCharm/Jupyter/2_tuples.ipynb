# 2. TUPLES — The Scroll of Ancient Coordinates

```python
coordinates = (5, 8, 12, 5, 42, 7, 8, 0, 99, 5)
```

```python
# Q1. Determine how many coordinates the scroll contains.

print(len(coordinates))
```

```python
# Q2. What is the first coordinate? The last one?

print(coordinates[0], coordinates[-1])
```

```python
# Q3. Slice out the first five coordinates.

print(coordinates[:5])
```

```python
# Q4. Extract only the even-positioned coordinates (0-based index).

print(coordinates[::2])
```

```python
# Q5. How many times does the number 5 appear?

print(coordinates.count(5))
```

```python
# Q6. Find the position (index) of the first occurrence of 42.

print(coordinates.index(42))
```

```python
# Q7. Unpack the first three coordinates into x, y, z and print them.

x, y, z = coordinates[:3]
print(x, y, z)
```

```python
# Q8. Use slicing and sum() to get the total of the middle five values.

print(sum(coordinates[2:7]))  # indices 2 to 6 = middle five
```

```python
# Q9. Convert the tuple into a list, and append the value 100.

coord_list = list(coordinates)
coord_list.append(100)
print(coord_list)
```

```python
# Q10. Sort the converted list in descending order.

coord_list.sort(reverse=True)
print(coord_list)
```

```python
# Q11. Convert it back to a tuple.

coord_tuple = tuple(coord_list)
print(coord_tuple)
```

```python
# Q12. Is the number 12 present in the original tuple?

print(12 in coordinates)
```

```python
# Q13. Count how many unique values are in the coordinates.

print(len(set(coordinates)))
```

```python
# Q14. Create a tuple that is the reverse of the original.

reversed_coords = coordinates[::-1]
print(reversed_coords)
```

```python
# Q15. Join two tuples: this one and (1, 2, 3) into a new tuple.

joined = coordinates + (1, 2, 3)
print(joined)
```

```python
# Q16. Repeat the coordinate tuple twice using repetition operator.

repeated = coordinates * 2
print(repeated)
```

```python
# Q17. Slice and extract coordinates from index 3 to 7.

print(coordinates[3:7])  # end index is exclusive; use [3:8] to include index 7
```

```python
# Q18. What is the minimum and maximum coordinate?

print(min(coordinates), max(coordinates))
```

```python
# Q19. Create a nested tuple: group first 3, next 3, and rest into sub-tuples.

nested = (coordinates[:3], coordinates[3:6], coordinates[6:])
print(nested)
```

```python
# Q20. Write a function that takes a tuple and returns only coordinates > 10.

def greater_than_10(t):
    return tuple(c for c in t if c > 10)

print(greater_than_10(coordinates))
```
