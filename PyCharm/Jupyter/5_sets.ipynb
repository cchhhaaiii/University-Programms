# 5. SETS — The Brotherhood of Unique Blades

```python
blades = {101, 204, 101, 305, 408, 204, 509, 612, 408}
```

```python
# Q1. Check how many unique blades are in the vault.

print(len(blades))
```

```python
# Q2. Add a new blade 713 to the collection.

blades.add(713)
print(blades)
```

```python
# Q3. Attempt to add 305 again — what happens?

blades.add(305)  # nothing happens: sets ignore duplicates
print(blades)
```

```python
# Q4. Remove blade 509 from the vault.

blades.remove(509)
print(blades)
```

```python
# Q5. Safely try removing blade 888 without causing an error.

blades.discard(888)  # no error even if missing
print(blades)
```

```python
# Q6. Create another set: new_findings = {305, 888, 999}

new_findings = {305, 888, 999}
print(new_findings)
```

```python
# Q7. Merge the two vaults into one (union).

merged = blades.union(new_findings)
print(merged)
```

```python
# Q8. Find the blades common to both vaults (intersection).

print(blades.intersection(new_findings))
```

```python
# Q9. Identify the blades only in the original set but not in new_findings.

print(blades.difference(new_findings))
```

```python
# Q10. Check if blade 204 is present in the vault.

print(204 in blades)
```

```python
# Q11. Convert the set to a list and sort it ascending.

blade_list = sorted(list(blades))
print(blade_list)
```

```python
# Q12. Find the largest and smallest blade numbers.

print(max(blades), min(blades))
```

```python
# Q13. Empty the set completely — the vault has been stolen!

blades.clear()
print(blades)
```

```python
# Q14. Restore it from backup: {101, 204, 305, 408, 509, 612}

blades = {101, 204, 305, 408, 509, 612}
print(blades)
```

```python
# Q15. Check if the vault is now a subset of another archive: {101, 204, 305, 408, 509,
#      612, 713, 888}

archive = {101, 204, 305, 408, 509, 612, 713, 888}
print(blades.issubset(archive))
```

```python
# Q16. Find if the archive is a superset of the vault.

print(archive.issuperset(blades))
```

```python
# Q17. Find the symmetric difference (blades that are in one but not both).

print(blades.symmetric_difference(archive))
```

```python
# Q18. Create a copy of the vault — you'll need it in the future.

blades_copy = blades.copy()
print(blades_copy)
```

```python
# Q19. Use a loop to print: "Blade #101 secured." for each blade.

for blade in sorted(blades):
    print(f"Blade #{blade} secured.")
```

```python
# Q20. Write a function that takes any set and returns only even-numbered blades.

def even_blades(s):
    return {b for b in s if b % 2 == 0}

print(even_blades(blades))
```
