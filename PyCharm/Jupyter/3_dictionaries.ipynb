# 3. DICTIONARIES — The Inventory of the Wizard

```python
inventory = {
    "wand": 3,
    "potion": 5,
    "scroll": 7,
    "elixir": 2,
    "gem": 10
}
```

```python
# Q1. The wizard needs to know how many types of items he owns.

print(len(inventory))
```

```python
# Q2. He wants to see only the names of the items.

print(list(inventory.keys()))
```

```python
# Q3. Now he needs just the item counts.

print(list(inventory.values()))
```

```python
# Q4. Show him both the item names and their counts.

print(list(inventory.items()))
```

```python
# Q5. A thief stole the elixir — remove it.

inventory.pop("elixir")
print(inventory)
```

```python
# Q6. He wants to know how many scrolls he has, but carefully — don't risk an error if
#      it's missing.

print(inventory.get("scroll", 0))
```

```python
# Q7. A new shipment arrives: {"crystal": 4, "wand": 2} — merge it into the inventory.

inventory.update({"crystal": 4, "wand": 2})
print(inventory)
```

```python
# Q8. He thinks "wand" should now be 10 — set it directly.

inventory["wand"] = 10
print(inventory)
```

```python
# Q9. Show whether the "dagger" exists in the inventory.

print("dagger" in inventory)
```

```python
# Q10. Show whether the "gem" is still safe.

print("gem" in inventory)
```

```python
# Q11. The wizard wants to reward you — print the most valuable item (highest count).

best = max(inventory, key=inventory.get)
print(best, inventory[best])
```

```python
# Q12. Also, show the item with the lowest quantity.

worst = min(inventory, key=inventory.get)
print(worst, inventory[worst])
```

```python
# Q13. He drops the last item added — remove it (even if you don't know its key).

print(inventory.popitem())
print(inventory)
```

```python
# Q14. Sort and show items by name (ascending).

print(dict(sorted(inventory.items())))
```

```python
# Q15. Sort and show items by value (descending).

print(dict(sorted(inventory.items(), key=lambda item: item[1], reverse=True)))
```

```python
# Q16. Loop through the inventory and print each item like: "wand: 3 uses".

for item, count in inventory.items():
    print(f"{item}: {count} uses")
```

```python
# Q17. Create a new dictionary where each quantity is doubled (without changing the
#      original).

doubled = {item: count * 2 for item, count in inventory.items()}
print(doubled)
print(inventory)  # original unchanged
```

```python
# Q18. Sum up the total number of items the wizard owns.

print(sum(inventory.values()))
```

```python
# Q19. The wizard forgets everything — clear the inventory.

backup = inventory.copy()  # saved copy for the next question
inventory.clear()
print(inventory)
```

```python
# Q20. But wait! He had a backup — restore it from a saved copy.

inventory = backup.copy()
print(inventory)
```
