```python
 #x+y+z=4 2x+y-z=1 x-y+2z=2
import numpy as np
A = np.array([
    [1,  1,  1],  # 1x + 1y + 1z
    [2,  1, -1],  # 2x + 1y - 1z
    [1, -1,  2]   # 1x - 1y + 2z
])

B = np.array([4, 1, 2])

solution = np.linalg.solve(A, B)

print(f"x = {round(solution[0], 3)}")
print(f"y = {round(solution[1], 3)}")
print(f"z = {round(solution[2], 3)}")
```
