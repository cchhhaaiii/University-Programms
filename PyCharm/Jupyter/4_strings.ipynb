# 4. STRINGS — The Tale of the Wise Coder

```python
story = " Once upon a TIME in a distant village, there lived a wise coder named Python. He was respected for his WISDOM, logic, and power. Python often said: 'Code is poetry!' "
```

```python
# Q1. The story feels cluttered — clean the unnecessary surroundings.

clean = story.strip()
print(clean)
```

```python
# Q2. The elders say Python should never be spoken in uppercase — make the name more
#      humble.

humble = clean.replace("Python", "python")
print(humble)
```

```python
# Q3. All wisdom must begin in quiet tone — make the whole story soft-spoken.

quiet = clean.lower()
print(quiet)
```

```python
# Q4. You're preparing a title for a scroll — turn it into a proper title.

print(clean.title())
```

```python
# Q5. The master hates loudness — remove every occurrence of "WISDOM" and replace it
#      with "clarity".

calm = clean.replace("WISDOM", "clarity")
print(calm)
```

```python
# Q6. Count how many times our master Python is mentioned (regardless of how it is
#      written).

print(clean.lower().count("python"))
```

```python
# Q7. Is it true that the story begins with silence? Prove it.

print(story.startswith(" "))  # the original begins with a blank space (silence)
```

```python
# Q8. Is there truth in the final words? Confirm if they end with a poetic mark.

print(clean.endswith("!'"))  # ends with the exclamation mark + closing quote
```

```python
# Q9. Extract the ancient quote Python said — it's between the single quotes.

start = clean.find("'") + 1
end = clean.rfind("'")
quote = clean[start:end]
print(quote)
```

```python
# Q10. Find where the word "power" first appears, and where the word "the" last
#      appears.

print(clean.find("power"))
print(clean.rfind("the"))  # matches the "the" inside "there"
```

```python
# Q11. The High Council wants a banner — take only the first 12 symbols.

print(clean[:12])
```

```python
# Q12. They also need the last 15 symbols for a secret seal.

print(clean[-15:])
```

```python
# Q13. There's a hidden code in the middle — extract every second character from
#      position 10 to 50.

print(clean[10:50:2])
```

```python
# Q14. To read the tale backward — reverse the story completely.

print(clean[::-1])
```

```python
# Q15. Break the story into its words, and count how many parts you get.

words = clean.split()
print(words)
print(len(words))
```

```python
# Q16. Some of the wisdom was stored in comma-separated runes — split those too.

runes = clean.split(",")
print(runes)
print(len(runes))
```

```python
# Q17. Join the words using a sacred rune: - (dash).

print("-".join(words))
```

```python
# Q18. Count how often the letter of enlightenment ('e') appears in the story.

print(clean.count("e"))
```

```python
# Q19. The scroll must begin with "Once" — verify that.

print(clean.startswith("Once"))
```

```python
# Q20. The scroll must end with "poetry!'" — check that too.

print(clean.endswith("poetry!'"))
```

```python
# Q21. There is a word "Code" — but is it pure? Is it alphabetic only? (Clean it first)

word = [w for w in words if "Code" in w][0]  # "'Code"
word = word.strip("'")
print(word, word.isalpha())
```

```python
# Q22. Check if a number like "123456" is considered holy — is it purely numeric?

print("123456".isnumeric())
```

```python
# Q23. Show how long the purified scroll is after trimming.

print(len(clean))
```

```python
# Q24. Write a code snippet that changes all " " (spaces) to underscores to preserve
#      spacing in ancient runes.

underscored = clean.replace(" ", "_")
print(underscored)
```

```python
# Q25. Finally, form a password using: the first 3 letters, the last 3 letters, and the
#      length of the final story.

password = clean[:3] + clean[-3:] + str(len(clean))
print(password)
```
